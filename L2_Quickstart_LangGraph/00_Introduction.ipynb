## LangGraph란?
LangGraph는 AI 에이전트와 애플리케이션을 위한 지속 가능한 런타임을 제공하는 프레임워크 

### [Why] 왜 Langgraph를 써야하나?
- Langchain 등을 통해 AI 애플리케이션을 시작하는 것은 쉽지만, 이를 맞춤화하고 확장하는 것은 어려움
  => 즉, 간단하게 실험실/연구실 단계에서 or 사내 배포 정도 레벨로 구현하는 건 쉽지만 실제 서비스 환경에서 AI 애플리케이션을 배포하고 관리하는 것이 어려움.
- 이를 위해 AI 애플리케이션 배포 과정에서 발생하는 주요 문제점들을 파악
    1. 지연시간 - Latency in the seconds vs ms
        - LangGraph는 LLM 모델을 더 빠르게 만들 수는 없지만, 두 가지 방법으로 개선할 수 있음
        1) 첫 번째는, `작업의 병렬화(Parallelization - to save actual latency)` : n개의 작업이 동시에 수행될 수 있다면 결론적으로 작업 시간을 줄일 수 있게 됨
        2) 두 번째는, `스트리밍(Streaming - to save perceived latency)` : 대화형 애플리케이션에서는 첫 번째 반환 토큰까지의 시간이 가장 중요한 지표인 경우가 많음
    2. 신뢰성 - Long-Running Agents can fail, whichi is expensive and time consuming
        - 수십 ms짜리 답변이라면 큰 상관이 없을 수도 있지만, 많은 자원이 들어갔거나 전체 컨택스트를 오염시킬 수 있는 답변이 있는 경우가 존재함
        1) `Checkpointing - to reduce the cost of each retry` : 각 단계에서 애플리케이션의 상태를 저장하고, 저장된 지점에서 다시 시작 가능
    3. LLM 응답의 비결정론적 특성 - The on-deterministic nature of AI requires checkpoints, approvals, and testing
        1) `Human-in-the-loop - to collaborate with the user` : 에이전트를 일시 중지하고 유저의 입력을 기다리는 유저 참여 인터페이스 구현
        2) `Tracing, Observation and Evaluation` by `LangSmith`
